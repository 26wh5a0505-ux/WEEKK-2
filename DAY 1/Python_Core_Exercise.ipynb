```python
import os
import datetime
import random

print("--- Intermediate Python Core Exercise ---")
print("This exercise covers Python fundamentals, data structures, functions,")
print("core libraries (os, datetime, random), file handling, and exception handling.")
print("Estimated completion time: 3-4 hours.")
print("------------------------------------------")


# --- Global Data for the Exercise ---
DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}


# ============================================================
# TASK 1
# ============================================================

print("\n--- Task 1: Data Generation and Basic Structures ---")


# 1.1 Generate 100 simulated sensor readings

raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(100):
    timestamp = start_time + datetime.timedelta(minutes=i)
    timestamp_str = timestamp.strftime('%Y-%m-%d %H:%M:%S')

    device_id = f"DEV{i + 1:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    minimum_value, maximum_value = READING_RANGES[device_type]
    value = round(random.uniform(minimum_value, maximum_value), 2)

    reading = (
        timestamp_str,
        device_id,
        device_type,
        location,
        value
    )

    raw_sensor_data.append(reading)

print("Number of readings:", len(raw_sensor_data))


# 1.2 Create temperature_readings

temperature_readings = []

for reading in raw_sensor_data:
    timestamp_str, device_id, device_type, location, value = reading

    if device_type == 'Temperature_Sensor':
        temperature_reading = {
            'timestamp': timestamp_str,
            'device_id': device_id,
            'location': location,
            'temperature_celsius': value
        }

        temperature_readings.append(temperature_reading)

print("Temperature readings:", len(temperature_readings))


# 1.3 Create unique_locations

unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print("Unique locations:", unique_locations)


# 1.4 Create device_stats

device_stats = {}

for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

for reading in raw_sensor_data:
    device_type = reading[2]
    value = reading[4]

    current_min, current_max = device_stats[device_type]

    if value < current_min:
        current_min = value

    if value > current_max:
        current_max = value

    device_stats[device_type] = (current_min, current_max)

print("Device statistics:")

for device_type, stats in device_stats.items():
    print(device_type, "->", stats)


# ============================================================
# TASK 1.5
# ============================================================

print("\n--- Task 1.5: Advanced Indexing and Slicing ---")

sample_string = "PythonProgrammingIsFun"


# 1.5.1 List slicing

sliced_readings_1 = raw_sensor_data[9:20]

sliced_readings_2 = raw_sensor_data[::5]

print("10th to 20th readings:", sliced_readings_1)
print("Every 5th reading:", sliced_readings_2)


# 1.5.2 String slicing

substring_1 = sample_string[6:17]

substring_2 = sample_string[5::-1]

print("Substring 1:", substring_1)
print("Substring 2:", substring_2)


# 1.5.3 Mixed slicing

fifth_reading_device_type = raw_sensor_data[4][2]

device_type_suffix = fifth_reading_device_type[-6:]

print("5th reading device type:", fifth_reading_device_type)
print("Device type suffix:", device_type_suffix)


# ============================================================
# TASK 2
# ============================================================

print("\n--- Task 2: Function Definitions and Usage ---")


# 2.1 Function to get readings by location

def get_readings_by_location(readings_list, target_location):
    result = []

    for reading in readings_list:
        if reading[3] == target_location:
            result.append(reading)

    return result


# 2.2 Function to calculate average temperature

def calculate_average_temperature(temp_data_list):
    if len(temp_data_list) == 0:
        return 0.0

    total_temperature = 0.0

    for reading in temp_data_list:
        total_temperature += reading['temperature_celsius']

    return total_temperature / len(temp_data_list)


# 2.3 Function to generate report line

def generate_report_line(reading_dict):
    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# Test Task 2 functions

lab1_readings = get_readings_by_location(raw_sensor_data, 'Lab1')
average_temperature = calculate_average_temperature(temperature_readings)

print("Lab1 readings:", len(lab1_readings))
print("Average temperature:", round(average_temperature, 2))

if len(temperature_readings) > 0:
    print("Sample report line:")
    print(generate_report_line(temperature_readings[0]))


# ============================================================
# TASK 3
# ============================================================

print("\n--- Task 3: Core Library Usage ---")


# 3.1 Create sensor_reports directory

reports_dir = 'sensor_reports'

if not os.path.exists(reports_dir):
    os.makedirs(reports_dir)
    print(f"Directory '{reports_dir}' created.")
else:
    print(f"Directory '{reports_dir}' already exists.")


# 3.2 Calculate duration between first and last timestamp

if len(raw_sensor_data) > 1:
    first_timestamp_str = raw_sensor_data[0][0]
    last_timestamp_str = raw_sensor_data[-1][0]

    date_format = '%Y-%m-%d %H:%M:%S'

    first_dt = datetime.datetime.strptime(
        first_timestamp_str,
        date_format
    )

    last_dt = datetime.datetime.strptime(
        last_timestamp_str,
        date_format
    )

    duration = last_dt - first_dt

    days = duration.days
    total_seconds = int(duration.total_seconds())

    hours = (total_seconds // 3600) % 24
    minutes = (total_seconds % 3600) // 60

    print(
        f"Duration of data collection: "
        f"{days} days, {hours} hours, {minutes} minutes."
    )
else:
    print("Not enough data to calculate duration.")


# 3.3 Generate 5 random integers

random_values = []

for i in range(5):
    random_values.append(random.randint(100, 200))

print("Random values:", random_values)


# ============================================================
# TASK 4
# ============================================================

print("\n--- Task 4: File I/O ---")


# 4.1 Write temperature readings to file

report_filename = os.path.join(
    reports_dir,
    'temperature_report.txt'
)

try:
    with open(report_filename, 'w', encoding='utf-8') as file:
        for reading_dict in temperature_readings:
            line = generate_report_line(reading_dict)
            file.write(line + '\n')

    print(f"Temperature report written to '{report_filename}'.")

except IOError as error:
    print(f"Error writing file '{report_filename}': {error}")


# 4.2 Read the report back

read_report_lines = []

try:
    with open(report_filename, 'r', encoding='utf-8') as file:
        read_report_lines = file.readlines()

    print("\nFirst 3 lines of the report:")

    for line in read_report_lines[:3]:
        print(line.strip())

except IOError as error:
    print(f"Error reading file '{report_filename}': {error}")


# ============================================================
# COMPLETE
# ============================================================

print("\n--- Exercise Complete! ---")
print("Great job completing the intermediate Python exercise.")
print("Review your code and outputs to ensure correctness and understanding.")
print("Consider adding more complex scenarios or user inputs for further practice.")
```
